````python
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import normalize
from google.colab import drive

drive.mount('/content/drive')

DATA_DIR = '/content/drive/MyDrive/MovieRec_Shared_Data/processed'

!ls "/content/drive/MyDrive/MovieRec_Shared_Data/processed"
````

Đọc và kiểm tra dữ liệu

````python
df = pd.read_csv(f"{DATA_DIR}/movies_catalog.csv")

df['clean_text'] = df['clean_text'].fillna('')

print("Tổng số phim:", len(df))
print("Phân chia dữ liệu:")
print(df['split'].value_counts())
print("Số văn bản rỗng:", (df['clean_text'] == '').sum())
````

Huấn luyện TF-IDF trên tập Train

````python
vectorizer = TfidfVectorizer(
    max_features=5000,
    ngram_range=(1, 2)
)

train_texts = df.loc[df['split'] == 'train', 'clean_text']

vectorizer.fit(train_texts)

print("Số đặc trưng TF-IDF:", len(vectorizer.vocabulary_))
````

Tạo vector cho toàn bộ phim

````python
tfidf_matrix = vectorizer.transform(df['clean_text'])

tfidf_norm = normalize(
    tfidf_matrix,
    norm='l2',
    axis=1
).toarray().astype(np.float32)

print("Kích thước vector:", tfidf_norm.shape)
````

Lưu kết quả

````python
save_path = f"{DATA_DIR}/tfidf_embeddings.npy"

np.save(save_path, tfidf_norm)

print("Đã lưu:", save_path)

# Kiểm tra lại
loaded = np.load(save_path)

print("Shape:", loaded.shape)
print("Data type:", loaded.dtype)
````
